 - Perform preprocessing of the data – change to lower case, remove stopwords using NLTK, and remove numbers.

```python
import nltk
import re
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

# Get English stopwords
stop_words = set(stopwords.words('english'))

def preprocess_text(text):
    # 1. Convert to lowercase
    text = str(text).lower()

    # 2. Remove numbers
    text = re.sub(r'\d+', '', text)

    # 3. Remove punctuation and non-alphabetic characters
    text = re.sub(r'\W+', ' ', text)

    # 4. Tokenize
    tokens = word_tokenize(text)

    # 5. Remove stopwords and short words
    tokens = [word for word in tokens if word not in stop_words and len(word) > 1]

    return tokens
```

#### Tokenise the data using word_tokenize from NLTK.

```python
# Apply preprocessing
google_df['processed'] = google_df['Comment'].apply(preprocess_text)
trustpilot_df['processed'] = trustpilot_df['Review Content'].apply(preprocess_text)
```

 - Find the frequency distribution of the words from each data set's reviews separately. You can use nltk.freqDist.

 - Plot a histogram/bar plot showing the top 10 words from each data set.

```python
from nltk.probability import FreqDist
import matplotlib.pyplot as plt

# Frequency distribution
def get_freq_dist(tokens_list):
    all_words = [word for tokens in tokens_list for word in tokens]
    word_freq = FreqDist(all_words)
    common_words = word_freq.most_common(10)
    return word_freq, common_words

def plot_top_words(freq_dist1=None, title1=None, freq_dist2=None, title2=None):
    plt.figure(figsize=(12, 5))

    if freq_dist1 is not None:
      plt.subplot(1, 2, 1)
      freq_dist1.plot(10, title=title1)

    if freq_dist2 is not None:
      plt.subplot(1, 2, 2)
      freq_dist2.plot(10, title=title2)

    plt.tight_layout()
    plt.show()
```

```python
google_freq, all_google_words = get_freq_dist(google_df['processed'])
trustpilot_freq, all_trustpilot_words = get_freq_dist(trustpilot_df['processed'])

plot_top_words(google_freq, 'Google Top 10 Words', trustpilot_freq, 'Trustpilot Top 10 Words')
```

# Conducting initial topic modelling:

With the data frame created in the previous step:

• Filter out the reviews that are from the locations common to both data sets.

• Merge the reviews to form a new list.



```python
# Find common locations (case-insensitive)
common_loc = set(google_df["Club's Name"].str.lower()) & set(trustpilot_df["Location Name"].str.lower())

# Filter reviews from common locations
google_common = google_df[google_df["Club's Name"].str.lower().isin(common_loc)]
trustpilot_common = trustpilot_df[trustpilot_df["Location Name"].str.lower().isin(common_loc)]

# Merge reviews into a single list
all_reviews = google_common['Comment'].tolist() + trustpilot_common['Review Content'].tolist()

print(f"Total number of common location reviews: {len(all_reviews)}")
```

Preprocess this data set.

Use BERTopic on this cleaned data set.

```python
!pip install bertopic[visualization]
from bertopic import BERTopic

preprocessed_reviews = [preprocess_text(review) for review in all_reviews]
```

Output: List out the top topics along with their document frequencies.

```python
# If preprocess returns a list of tokens, convert to strings:
preprocessed_reviews = [' '.join(tokens) if isinstance(tokens, list) else str(tokens) for tokens in preprocessed_reviews]

# Topic Model
topic_model = BERTopic(language="english", calculate_probabilities=True, verbose=True)
topics, probs = topic_model.fit_transform(preprocessed_reviews)

# 1. List top topics with document frequencies
topic_info = topic_model.get_topic_info()
print("\nTop Topics and Document Frequencies:")
print(topic_info[['Topic', 'Count', 'Name']])
```

For the top 2 topics, list out the top words.


```python
# List top words for the top 2 topics
print("\nTop Words for Top 2 Topics:")
for topic_id in topic_info['Topic'].head(2):  # Top 2 topics
    if topic_id != -1:  # Skip outlier topic
        words = topic_model.get_topic(topic_id)
        print(f"\nTopic {topic_id}:")
        print([word[0] for word in words[:5]])  # Top 5 words
```

```python
import nltk
import gensim
import string
from nltk.corpus import stopwords
from nltk.stem.wordnet import WordNetLemmatizer
import pyLDAvis.gensim_models as gensimvis
import pyLDAvis
from gensim.models.ldamodel import LdaModel

nltk.download('wordnet')

# # Function to clean the document.
# def clean(doc):
#     # Remove stop words and convert to lowercase.
#     stop_free = " ".join([word for word in doc.lower().split() if word not in stop])
#     # Remove punctuation.
#     punc_free = ''.join(ch for ch in stop_free if ch not in exclude)
#     # Lemmatise the text.
#     normalized = " ".join(lemma.lemmatize(word) for word in punc_free.split())
#     return normalized

# Lowercase, remove punctuation, tokenize, remove stopwords
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess_LDA(text):
    text = text.lower()
    text = re.sub(r'\d+', '', text)
    text = text.translate(str.maketrans('', '', string.punctuation))
    tokens = word_tokenize(text)
    tokens = [word for word in tokens if word not in stop_words and len(word) > 2]
    # Lemmatize
    tokens = [lemmatizer.lemmatize(word) for word in tokens]
    return tokens

# Combine your negative reviews
all_negative_reviews = google_negative['Comment'].tolist() + trustpilot_negative['Review Content'].tolist()

# Create a dictionary representation of the documents.
tokenized_reviews = [preprocess_LDA(review) for review in all_negative_reviews]
```

```python
from gensim import corpora
dictionary = corpora.Dictionary(tokenized_reviews)

# Filter out words that occur fewer than 2 documents or more than 50% of the documents.
dictionary.filter_extremes(no_below=2, no_above=0.5)
```

- Build the LDA model using the dictionary and corpus

```python
# Create a BOW representation of the documents.
corpus = [dictionary.doc2bow(doc) for doc in tokenized_reviews]

# Set parameters.
num_topics = 10
passes = 20
```

- Create an LDA Model

```python
# Create the LDA model.
from gensim.models.ldamodel import LdaModel
lda_model = LdaModel(corpus=corpus, num_topics=num_topics, id2word=dictionary, passes=passes)
```
