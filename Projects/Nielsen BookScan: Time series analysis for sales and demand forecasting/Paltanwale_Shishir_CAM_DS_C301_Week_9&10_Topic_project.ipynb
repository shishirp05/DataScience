### ISBN Resample function

```python
def resample_isbn(group):
    try:
        # Check if group is empty
        if group is None or group.empty:
            print("⚠️ Empty or None group detected")
            return pd.DataFrame()

        # Ensure End_Date column exists
        if 'End_Date' not in group.columns:
            print(f"⚠️ 'End_Date' missing in group for ISBN {group['ISBN'].iloc[0]}")
            return pd.DataFrame()

        # Ensure datetime type
        group['End_Date'] = pd.to_datetime(group['End_Date'], errors='coerce')

        # Drop rows with invalid dates
        group = group.dropna(subset=['End_Date'])

        if group.empty:
            print(f"⚠️ All 'End_Date' values are NaT for ISBN {group['ISBN'].iloc[0]}")
            return pd.DataFrame()

        # Set index
        group = group.set_index('End_Date')
        start = group.index.min()
        end = group.index.max()

        # Build full weekly index
        full_index = pd.date_range(start=start, end=end, freq='W')

        # Resample
        weekly = group.resample('W').agg({
            'Volume': 'sum',
            'Value': 'sum',
            'ASP': 'mean',
            'Title': 'first',
            'Author': 'first',
            'RRP': 'first',
            'Binding': 'first',
            'Imprint': 'first',
            'Publisher_Group': 'first',
            'Product_Class': 'first'
        }).reindex(full_index)

        # Fill missing numeric fields
        weekly['Volume'] = weekly['Volume'].fillna(0)
        weekly['Value'] = weekly['Value'].fillna(0)
        weekly['ASP'] = weekly['ASP'].fillna(0)

        # Fill string metadata
        for col in ['Title', 'Author', 'RRP', 'Binding', 'Imprint', 'Publisher_Group', 'Product_Class']:
            weekly[col] = weekly[col].fillna(method='ffill').fillna(method='bfill')

        # Add ISBN back
        weekly['ISBN'] = group['ISBN'].iloc[0]

        # Rename index
        weekly.index.name = 'End_Date'

        return weekly.reset_index()

    except Exception as e:
        print(f"❌ Error processing ISBN group:\n{e}\nGroup sample:\n{group.head()}")
        return pd.DataFrame()
```

```python
# Create the input-output pairs in the format required for time series forecasting.
# This function creates sequences where the input is 'lookback' steps
# and the output is 'forecast' steps.
def create_input_output_sequences_xgb(lookback, forecast, sequence_data):
  input_sequences = []
  output_sequences = []

  for i in range(lookback, len(sequence_data) - forecast + 1):

      input_sequences.append(sequence_data[i - lookback: i])
      output_sequences.append(sequence_data[i: i + forecast])

  return input_sequences, output_sequences
  # return np.array(input_sequences), np.array(output_sequences)
```

#### Best Parameters function

```python
from hyperopt import fmin, tpe, Trials

def run_hyperopt_tuning(train_df, test_df, param_space, lookback, horizon, max_evals=100):
    """
    Runs Hyperopt tuning using the provided training/testing data and parameter space.

    Parameters:
    - train_df: pd.DataFrame, training data
    - test_df: pd.DataFrame, testing data
    - param_space: dict, Hyperopt search space
    - max_evals: int, number of evaluations (default: 100)

    Returns:
    - best_params: dict, best hyperparameters found
    - trials: hyperopt.Trials object, contains all trial results
    """
    trials = Trials()

    best_params = fmin(
        fn=lambda params: auto_tune_training(params, train_df, test_df, lookback, horizon),
        space=param_space,
        algo=tpe.suggest,
        max_evals=max_evals,
        trials=trials
    )

    return best_params, trials
```

```python
# Function for setting up the hyperparameter optimisation.
def auto_tune_training(param_tuning_obj, train_set, test_set, lookback, horizon):

  # lookback = 64
  # horizon = 32

  train_input, train_output = create_input_output_sequences_xgb(lookback, horizon, train_set)
  test_input, test_output = create_input_output_sequences_xgb(lookback, horizon, test_set)

  # if len(train_input) != 0:
  train_input = np.array(train_input).reshape(len(train_input), -1)

  # if len(test_input) != 0:
  test_input  = np.array(test_input).reshape(len(test_input), -1)

  train_output = np.array(train_output)
  test_output  = np.array(test_output)

  # If output has 3 dims, reshape to 2 dims
  if train_output.ndim == 3:
      train_output = train_output.reshape(train_output.shape[0], train_output.shape[1])
  if test_output.ndim == 3:
      test_output = test_output.reshape(test_output.shape[0], test_output.shape[1])

  model = XGBRegressor(
          n_estimators=param_tuning_obj['n_estimators'],
          max_depth=int(param_tuning_obj['max_depth']),
          gamma=param_tuning_obj['gamma'],
          reg_alpha=int(param_tuning_obj['reg_alpha']),
          min_child_weight=int(param_tuning_obj['min_child_weight']),
          colsample_bytree=float(param_tuning_obj['colsample_bytree']),
          eval_metric="rmse",
          early_stopping_rounds=10
        )

  evaluation = [(train_input, train_output), (test_input, test_output)]

  model.fit(train_input, train_output,
            eval_set=evaluation,
            verbose=False)

  pred = model.predict(test_input)
  accuracy = mean_absolute_error(test_output, pred)
  print("MAE:", accuracy)
  return {'loss': accuracy, 'status': STATUS_OK }
```

```python
# Create a pipeline with detrender and deseasonaliser.
def create_predictor_with_deseasonaliser_xgboost(sp=12, degree=1, best_xgb_params=None):

    # regressor = XGBRegressor(base_score=0.5,
    #                   n_estimators=400,
    #                   min_child_weight=1,
    #                   max_depth=7,
    #                   learning_rate=0.1,
    #                   booster='gbtree',
    #                   tree_method='exact',
    #                   reg_alpha=0,
    #                   subsample=0.5,
    #                   validate_parameters=1,
    #                   colsample_bylevel=1,
    #                   colsample_bynode=1,
    #                   colsample_bytree=1,
    #                   gamma=0
    #                 )

    if best_xgb_params is None:
        best_xgb_params = {
            'base_score': 0.5,
            'n_estimators': 400,
            'min_child_weight': 1,
            'max_depth': 7,
            'learning_rate': 0.1,
            'booster': 'gbtree',
            'tree_method': 'exact',
            'reg_alpha': 0,
            'subsample': 0.5,
            'validate_parameters': 1,
            'colsample_bylevel': 1,
            'colsample_bynode': 1,
            'colsample_bytree': 1,
            'gamma': 0
        }

    regressor = XGBRegressor(**best_xgb_params)

    forecaster = TransformedTargetForecaster(
        [
            ("deseasonalize", Deseasonalizer(model="additive", sp=sp)), # (model="multiplicative", sp=sp)),
            ("detrend", Detrender(forecaster=PolynomialTrendForecaster(degree=degree))),
            (
                "forecast",
                make_reduction(regressor, window_length=12, strategy="recursive"),
            ),
        ]
    )

    return forecaster
```

```python
# Perform a grid search with cross-validation.
def grid_search_predictor(train, test, predictor, param_grid):

    # Grid search on window_length.
    cv = ExpandingWindowSplitter(initial_window=int(len(train) * 0.7))

    gscv = ForecastingGridSearchCV(
        predictor, strategy="refit", cv=cv, param_grid=param_grid,
        scoring=MeanAbsolutePercentageError(symmetric=True),
        error_score='raise'
    )

    gscv.fit(train)
    print(f"Best parameters: {gscv.best_params_}")

    # Forecast.
    future_horizon = np.arange(1, 33) #np.arange(len(test)) + 1
    predictions = gscv.predict(fh=future_horizon)

    return gscv.best_params_, predictions
```

```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd

def prepare_lstm_sequences(df, isbn, start_date='2012-01-01', forecast_horizon=32):
    """
    Prepares LSTM-ready sequences for a given book.

    Parameters
    ----------
    df : pd.DataFrame
        Weekly sales data containing 'ISBN', 'End_Date', 'Volume'.
    isbn : str
        ISBN of the book to filter.
    start_date : str, optional
        Start date for filtering data, by default '2012-01-01'.
    forecast_horizon : int, optional
        Number of weeks for the forecast horizon, by default 32.

    Returns
    -------
    train_seq : np.ndarray
        Scaled training sequence.
    test_seq : np.ndarray
        Scaled test sequence (forecast horizon).
    scaler : sklearn.preprocessing.MinMaxScaler
        Fitted scaler, useful for inverse transforming predictions.
    """
    # Filter for the book and date
    book_df = df[(df['ISBN'] == isbn) & (df['End_Date'] > start_date)].sort_values('End_Date')

    # Scale the 'Volume' column
    scaler = MinMaxScaler(feature_range=(0, 1))
    seq_data = scaler.fit_transform(book_df[['Volume']])

    # Split into train/test based on forecast horizon
    train_seq = seq_data[:-forecast_horizon]
    test_seq  = seq_data[-forecast_horizon:]

    return book_df, train_seq, test_seq, scaler
```

```python
def prepare_lstm_input(df_seq_data, lookback, forecast):
    """
    Prepare LSTM input and output sequences and get the last available input sequence for forecasting.

    Parameters:
    -----------
    df_seq_data : np.array or pd.DataFrame
        Scaled sequence data (e.g., 'Volume').
    lookback : int
        Number of past timesteps to use as input.
    forecast : int
        Number of future timesteps to predict.

    Returns:
    --------
    input_seq : np.array
        LSTM input sequences, shaped (samples, lookback, 1)
    output_seq : np.array
        LSTM output sequences, shaped (samples, forecast)
    last_input_seq : np.array
        Last input sequence to predict the next `forecast` timesteps, shaped (1, lookback, 1)
    """
    # Create sequences
    input_seq, output_seq = create_input_output_sequences_xgb(lookback, forecast, df_seq_data)
    input_seq = np.array(input_seq)
    output_seq = np.array(output_seq)

    # Reshape for LSTM
    input_seq = input_seq.reshape(input_seq.shape[0], input_seq.shape[1], 1)

    # Last available sequence for prediction
    last_input_seq = input_seq[-1].reshape(1, lookback, 1)

    return input_seq, output_seq, last_input_seq
```

#### Function to create Tuned LSTM Model

```python
# Define the model for hyperparameter tuning.
from tensorflow import keras
from keras import layers
import keras_tuner as kt

def tuned_model(hp, lookback=None):
    model = keras.Sequential()
    model.add(LSTM(hp.Int('input_unit',min_value=4,max_value=128,step=8),return_sequences=True, input_shape=(lookback, 1))) # 12 == lookback

    for i in range(hp.Int('n_layers', 1, 4)):
        model.add(LSTM(hp.Int(f'lstm_{i}_units',min_value=4,max_value=128,step=8),return_sequences=True))

    model.add(LSTM(hp.Int('layer_2_neurons',min_value=4,max_value=128,step=8)))
    model.add(Dropout(hp.Float('Dropout_rate',min_value=0,max_value=0.5,step=0.1)))
    model.add(Dense(1, activation=hp.Choice('dense_activation',values=['relu', 'sigmoid'],default='relu'))) # 1 == forecast
    model.compile(loss='mean_squared_error', optimizer='adam',metrics = ['mse'])

    return model
```

```python
import pmdarima as pm

from statsmodels.tsa.statespace.sarimax import SARIMAX

def fit_sarimax_model(endog, exog=None, order=(1, 1, 1), seasonal_order=(1, 0, 1, 12), trend='c', maxiter=500, disp=False):
    """
    Fits a SARIMAX model to a time series with optional exogenous variables.

    Parameters:
    - endog: pandas Series, the target time series
    - exog: pandas DataFrame or Series, optional exogenous variables
    - order: tuple (p, d, q), ARIMA order
    - seasonal_order: tuple (P, D, Q, m), seasonal ARIMA order
    - trend: str, trend specification ('n', 'c', 't', 'ct')
    - maxiter: int, maximum number of iterations for fitting
    - disp: bool, whether to display convergence output

    Returns:
    - results: fitted SARIMAX model results object
    """
    model = SARIMAX(endog=endog,
                    exog=exog,
                    order=order,
                    seasonal_order=seasonal_order,
                    trend=trend)

    results = model.fit(maxiter=maxiter, disp=disp)
    print(results.summary())
    return model, results
```

```python
def create_lag_features(y, lag=1):
    df = pd.DataFrame(y)
    for i in range(1, lag + 1):
        df[f'lag_{i}'] = df[0].shift(i)

    df.dropna(inplace=True)  # Drop rows with NaN values.
    return df
```

```python
import numpy as np

def prepare_hybrid_forecast_data(
    alpha,
    lstm_forecast,
    sarima_forecast,
    df_hybrid,
    forecast_horizon
):
    """
    Combines LSTM and SARIMA forecasts using a weighted average and prepares data for visualization.

    Parameters:
    - alpha: float, weight for LSTM (0 to 1)
    - lstm_forecast: array-like, LSTM forecast values
    - sarima_forecast: array-like, SARIMA forecast values
    - df_hybrid: DataFrame with historical 'Volume' data
    - forecast_horizon: int, number of forecast steps

    Returns:
    - hybrid_output: DataFrame ready for plotting (with 'Actual' and 'Forecast' columns)
    """

    # Combine forecasts
    hybrid_pred = alpha * np.array(lstm_forecast) + (1 - alpha) * np.array(sarima_forecast)

    # Prepare output for visualization
    hybrid_output = data_for_visualisations(df_hybrid, hybrid_pred, forecast_horizon)

    return hybrid_output, hybrid_pred
```

#### Function to plot the hybrid data

```python
from sklearn.metrics import mean_absolute_error, mean_absolute_percentage_error

def evaluate_and_plot_forecast(df, forecast_horizon, predictions, data_visualisation, plot_title=None):
    """
    Plots results and calculates MAE and MAPE for the given forecast.

    Parameters
    ----------
    df : pd.DataFrame
        DataFrame containing at least a 'Volume' column.
    forecast_horizon : int
        Number of periods in the forecast horizon (e.g., 32 for 32 weeks).
    predictions : array-like
        Model forecast values for the horizon.
    plot_func : callable, optional
        Function to plot results. Should accept (df, predictions, title).
    plot_title : str, optional
        Title for the plot.
    """
    # Step 1: Get actual values (last N periods)
    actual = df['Volume'][-forecast_horizon:]

    # Step 2: Calculate metrics
    mae = mean_absolute_error(actual, predictions)
    mape = mean_absolute_percentage_error(actual, predictions)

    # Step 3: Plot if a plot function is provided
    # if plot_func is not None:
    plot_with_secondary_axis(data_visualisation, plot_title)

    # Step 4: Print metrics
    print(f"MAE: {mae:.2f}")
    print(f"MAPE: {mape:.2%}")

    return mae, mape
```
