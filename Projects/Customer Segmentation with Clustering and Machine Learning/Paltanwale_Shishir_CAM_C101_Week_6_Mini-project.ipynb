# 3. Perform feature engineering as follows:
  - Create new features for frequency, recency, CLV, average unit cost, and customer age.
  - Perform feature scaling and encoding if needed.

```python
# Add column 'Frequency'
df['Frequency' ] = df.groupby( 'Customer ID' )['Order ID' ].transform( 'count')

# Calculate new column 'Recency' (time since last order)
df['Recency'] = (pd.to_datetime('today') - pd.to_datetime(df['Delivery_Date'])).dt.days

# Add column CLV
df['CLV'] = df['Total Revenue'].str.replace('$', '', regex=False)
df['CLV'] = pd.to_numeric(df['CLV'], errors='coerce')

# Add column Average Unit Cost
df['Average Unit Cost'] = df['Unit Cost'].str.replace('$', '', regex=False)
df['Average Unit Cost'] = pd.to_numeric(df['Average Unit Cost'], errors='coerce')

# Convert the DOB to datetime
df[ 'DOB'] = pd.to_datetime(df[ 'Customer_BirthDate' ], format= '%d%b%Y')

# Calculate age
today = pd.Timestamp( 'now')
df[ 'Customer Age'] = (today - df[ 'DOB']).dt.days // 365

print(df.shape)
```

### Aggregate the data into one 'Customer' in per row

```python
df = df.drop_duplicates(keep='first')
#df = df[['Customer ID']].drop_duplicates()
print(df.shape)

aggregated_df = df.groupby('Customer ID').agg({
    'Frequency': 'sum',
    'Recency': 'max',
    'CLV': 'sum',
    'Average Unit Cost': 'mean',
    'Customer Age': 'first'
}).reset_index()

print(aggregated_df.shape)
aggregated_df.head(total_rec)
```

# 5. For more efficient programming, incorporate column transformer and pipeline.

```python
# Drop null values for NON SCALED DATA
df_preproc = df_filtered.dropna()
df_preproc = df_preproc.drop('Outlier', axis=1)

# Pipeline for preprocessing
preprocessor = ColumnTransformer(transformers=[('num', StandardScaler(), df_columns)])

# Fit and transform the data NON SCALED data
df_preprocessed = preprocessor.fit_transform(df_preproc)

# Initialize the K-Means algorithm - the number of clusters (n_clusters) will need to be determined
kmeans = KMeans(n_clusters= 4, random_state=42, n_init=10)

# Create a pipeline that first preprocesses the data and then applies K-Means
pipeline = Pipeline(steps=[( 'preprocessor', preprocessor),( 'clusterer', kmeans)])

# Fit the pipeline
pipeline.fit(df_preproc)

# Get the cluster assignments
cluster_labels = pipeline.predict(df_preproc)

# Add cluster labels to the DataFrame
df_cluster = df_preproc.copy()
df_cluster['Cluster'] = cluster_labels

#tot_clusters = len(df_cluster['Cluster'].unique())

print(cluster_labels)
#print("Total clusters:\n", tot_clusters)

#print("Cluster Centers (Standardized):")
#print(pd.DataFrame(kmeans.cluster_centers_, columns=df_columns))

#print("\nClusters")
#print(df_cluster[df_cluster['Cluster'] == 1].sum)
```

# 6. Select the optimum value of clusters ( 𝑘 ) with the Elbow and Silhouette score methods.

## Elbow method and Silhouette score

```python
# Define a range of k values
k_values = range(2, 15)
wcss = []
# inertia = []
silhouette_scores = []

for k in k_values:
    #kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    #labels = kmeans.fit_predict(df_scaled)
    #wcss.append(kmeans.inertia_)
    #silhouette_scores.append(silhouette_score(df_scaled, labels))


    pipeline.set_params(clusterer__n_clusters=k)
    pipeline.fit(df_preproc)
    # Append WCSS (inertia)
    wcss.append(pipeline.named_steps['clusterer'].inertia_)
    # Compute silhouette score (skip for k=1, requires >1 cluster)
    labels = pipeline.named_steps['clusterer'].labels_
    silhouette_scores.append(
        silhouette_score(pipeline.named_steps['preprocessor'].transform(df_preproc), labels)
    )

# Create Side-by-side plots
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# Plot Elbow Method
ax[0].plot(k_values, wcss, marker='o', linestyle='-', color='red')
ax[0].set_xlabel("Number of Clusters (k)")
ax[0].set_ylabel("WCSS (Inertia)")
ax[0].set_title("Elbow Method for Optimal k")

# Plot Silhouette Score
ax[1].plot(k_values, silhouette_scores, marker='o', linestyle='-', color='orange')
ax[1].set_xlabel("Number of Clusters (k)")
ax[1].set_ylabel("Silhouette Score")
ax[1].set_title("Silhouette Score for Optimal k")

# Plot Elbow Curve
#plt.figure(figsize=(8, 5))
#plt.plot(k_values, wcss, marker='o', label='WCSS (Elbow)')
#plt.title('Elbow Method')
#plt.xlabel('Number of Clusters (k)')
#plt.ylabel('WCSS (Inertia)')
#plt.legend()
#plt.grid()
#plt.show()
```

**Elbow Method**

In the Elbow method, the curve does not show a very sharp bend, making it somewhat smooth and ambiguous. However, based on the visual inspection of the curve, 6 to 8 clusters appear to be a reasonable choice for segmenting the data.

**Silhouette Score**

Silhouette scores range from -1 to 1, with higher values indicating better-defined and more distinct clusters. A score near 1 means points are well-clustered, while values near -1 suggest poor clustering. In this analysis, the highest score is 0.28 for 1 cluster, and the lowest is 0.21 for 14 clusters. Scores around 0.25 are observed for 4 and 6 clusters, while 5 clusters fall slightly higher between 0.25 and 0.26. Based on these values, it can be reasonably concluded that the optimal number of clusters likely falls between 4 and 6.

# Silhouette plots

```python
# Perform PCA to reduce dimensions
silh_pca = PCA(n_components=2)
df_preproc_array = silh_pca.fit_transform(df_scaled)
#df_preproc_array = np.array(df_preproc)

k_values = range(2, 8)
for n_clusters in k_values:
    # Create a subplot with 1 row and 2 columns.
    fig, (ax1, ax2) = plt.subplots(1, 2)
    fig.set_size_inches(18, 7)

    # The 1st subplot is the silhouette plot.
    # The silhouette coefficient can range from -1, 1 but in this example all
    # lie within [-0.1, 1].
    ax1.set_xlim([-0.1, 1])
    # The (n_clusters+1)*10 is for inserting blank space between silhouette
    # plots of individual clusters, to demarcate them clearly.
    ax1.set_ylim([0, len(df_preproc_array) + (n_clusters + 1) * 10])

    # Initialize the clusterer with n_clusters value and a random generator
    # seed of 10 for reproducibility.
    clusterer = KMeans(n_clusters=n_clusters, random_state=10, n_init = 10)
    cluster_labels = clusterer.fit_predict(df_preproc_array)

    # The silhouette_score gives the average value for all the samples.
    # This gives a perspective into the density and separation of the formed
    # clusters
    silhouette_avg = silhouette_score(df_preproc_array, cluster_labels)
    print("For n_clusters =", n_clusters,
          "The average silhouette_score is :", silhouette_avg)

    # Compute the silhouette scores for each sample
    sample_silhouette_values = silhouette_samples(df_preproc_array, cluster_labels)

    y_lower = 10
    for i in range(n_clusters):
        # Aggregate the silhouette scores for samples belonging to
        # cluster i, and sort them.
        ith_cluster_silhouette_values = \
            sample_silhouette_values[cluster_labels == i]

        ith_cluster_silhouette_values.sort()

        size_cluster_i = ith_cluster_silhouette_values.shape[0]
        y_upper = y_lower + size_cluster_i

        color = cm.nipy_spectral(float(i) / n_clusters)
        ax1.fill_betweenx(np.arange(y_lower, y_upper),
                          0, ith_cluster_silhouette_values,
                          facecolor=color, edgecolor=color, alpha=0.7)

        # Label the silhouette plots with their cluster numbers at the middle.
        ax1.text(-0.05, y_lower + 0.5 * size_cluster_i, str(i))

        # Compute the new y_lower for next plot.
        y_lower = y_upper + 10  # 10 for the 0 samples.

    ax1.set_title("The silhouette plot for the various clusters.")
    ax1.set_xlabel("The silhouette coefficient values")
    ax1.set_ylabel("Cluster label")

    # The vertical line for average silhouette score of all the values.
    ax1.axvline(x=silhouette_avg, color="red", linestyle="--")

    ax1.set_yticks([])  # Clear the yaxis labels / ticks.
    ax1.set_xticks([-0.1, 0, 0.2, 0.4, 0.6, 0.8, 1])

    # 2nd Plot showing the actual clusters formed.
    colors = cm.nipy_spectral(cluster_labels.astype(float) / n_clusters)
    ax2.scatter(df_preproc_array[:, 0], df_preproc_array[:, 1], marker='.', s=30, lw=0, alpha=0.7,
                c=colors, edgecolor='k')

    # Labeling the clusters.
    centers = clusterer.cluster_centers_
    # Draw white circles at cluster centers.
    ax2.scatter(centers[:, 0], centers[:, 1], marker='o',
                c="white", alpha=1, s=200, edgecolor='k')

    for i, c in enumerate(centers):
        ax2.scatter(c[0], c[1], marker='$%d$' % i, alpha=1,
                    s=50, edgecolor='k')

    ax2.set_title("The visualization of the clustered data.")
    ax2.set_xlabel("Feature space for the 1st feature")
    ax2.set_ylabel("Feature space for the 2nd feature")

    plt.suptitle(("Silhouette analysis for KMeans clustering on sample data "
                  "with n_clusters = %d" % n_clusters),
                 fontsize=14, fontweight='bold')

plt.show()

```

Cluster 7, although fairly evenly distributed, contains a higher number of negative Silhouette values compared to the others, suggesting weaker cluster assignment for some data points.

Overall, Cluster 6 demonstrates better structure and separation than Clusters 5 and 7.
